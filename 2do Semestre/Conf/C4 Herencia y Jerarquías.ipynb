# 4 HERENCIA Y JERARQUÍAS DE TIPOS
La **herencia** es un recurso de programación que promueve la expresividad, la reusabilidad y la flexibilidad.

Consideremos que queremos trabajar con diferentes tipos de animales que tienen funciones en común pero con la especificidad de cada tipo de animal. La clase `Animal` de más abajo nos dice que todo animal tiene un `nombre` (un string) y emite un sonido (función `mi_sonido`). La definición de `__str__` nos expresa cómo presentar al animal (en este caso diciendo de qué tipo es, cuál es su nombre y que sonido hace). En el siguiente código hay cuatro clases: Perro, Gato, Pato y Leon. Que todas estas clases tienen en común lo que se describe en la clase Animal se expresa con el concepto de herencia. Cuando se escribe Perro(Animal) se está diciendo que la definición de Perro hereda de (proviene de, se deriva de) de la clase Animal. No hay que volver a escribir en el código de cada clase lo que ya fue descrito en la clase animal.

Por heredar de la clase Animal un objeto de tipo Perro tiene un nombre, tiene una forma de escribirse como string y emite un sonido simbolizado por el string que puede devolver la función mi_sonido. Por ello cada clase ha implementado de manera diferente la función mi_sonido. Pero todos los objetos sean Perro, Gato, Pato o Leon han implementado una función del mismo nombre mi_sonido. La implementación del __str__ incluida en Animal no tiene que reescribirse en la clases que heredan de Animal porque todos tienen una variable nombre, se les puede aplicar la función type y suponemos que una implementación de mi_sonido.

De modo que si se hace

`pe = Perro("Pluto")
print(pe)`

se usará el `__str__ `definido en `Animal` ya que `Perro` hereda de `Animal`, cuando al ejecutar este `__str__ `se hace

`type(self).__name__`

se devuelve el string `Perro` que es el nombre del tipo, luego `self.nombre` nos da el valor de la variable nombre que tendrá el nombre del animal y la función `mi_sonido` nos debe devolver el sonido que hace dicho animal, en este caso un `Perro`


```python
import math


class Animal:
    def __init__(self, nombre: str):
        self.nombre = nombre

    def __str__(self):
        return f"Soy un {type(self).__name__} de nombre {self.nombre} que hace {self.mi_sonido()}"

    def mi_sonido(self):
        pass


#Al no definir un __init__ se llama implicitamente al de la clase base (super)
class Perro(Animal):
    def mi_sonido(self):
        return "Jau, Jau"


class Gato(Animal):
    def mi_sonido(self):
        return "Miau, Miau"


class Pato(Animal):
    def mi_sonido(self):
        return "Cuac, Cuac"


class Leon(Animal):
    def mi_sonido(self):
        return "Gouarr, Gouarr"


pe = Perro("Pluto")
print(pe)
g = Gato("Tom")
print(g)
pa = Pato("Donald")
print(pa)
l = Leon("Simba")
print(l)
a = Animal(None)
print(a)
jauria = [pe, g, pa, l]
print(jauria)
#Lo que quisiéramos es no poder crear un Animal a secas sino que tenga que ser de una de las clases herederas
```

El código

`def a_sonar(jauria: List[Animal]) -> List[str]:
    return [a.mi_sonido() for a in jauria]`

`print(a_sonar(jauria))`

nos escribe los sonidos que hace cada uno de los animales de la lista `jauria` porque al hacer `a.mi_sonido()` aplica el `mi_sonido` implementado en cada animal

```python
from typing import List

def a_sonar(jauria: List[Animal]) -> List[str]:
    return [a.mi_sonido() for a in jauria]

print(a_sonar(jauria))
```

El código a continuación define otra clase `Alien` que también hereda de `Animal` pero que no implementa su propio `mi_sonido`. En este caso entonces la función `a_sonar` cuando aplique `mi_sonido` al objeto de tipo `Alien` aplicará el que éste hereda de `Animal` y que no hará nada (`pass`)no emitirá nada por lo que Python escribirá `None`

```python
class Alien(Animal):
    pass


pe = Perro("Pluto")
print(pe)
g = Gato("Tom")
print(g)
pa = Pato("Donald")
print(pa)
l = Leon("Simba")
print(l)
a = Animal(None)
print(a)
jauria = [pe, g, pa, l]
print(a_sonar(jauria))
jauria.append(Alien("Predator"))
print("\nDejé pasar que metieras un Alien en la jauría")
print(a_sonar(jauria))  #Dará un None en el caso de Animal

#El error debería ser por no permitir crear un animal que no implemente mi_sonido
```

## 4.1 Métodos abstractos
En el código anterior estamos en presencia de una situación típica de muchas jerarquías de tipo que heredan de una clase común: sabemos que todas tienen implementar un método del mismo nombre (en este ejemplo `mi_sonido`) pero que en la case **base** (padre) no se sabe como implementarlo, y por eso se le puso `pass `al código del método. Esto es lo que se llama **método abstracto** en la jerga de la POO, se sabe que debe existir, que es un método común para todas las clases que hereden de esta clase (`Animal` en este ejemplo) pero que debe implementarse en cada una de las clases que hereden de esta clase base.

En Python hay una mejor manera de especificar esto y es definiendo el método con el decorador `@abstractmethod` precediendo la definición. Para usar este recurso de Python hay que importar del módulo `abc` la clase `ABC` y el decorador `abstractmethod`. Ver código a continuación.

¿Qué pasa cuando queremos poner un objeto de tipo `Alien` (que tampoco implementa `mi_sonido`) en la lista jauria?. Si en el código de la celda anterior no pasaba nada cuando hicimos

`jauria.append(Alien("Predator"))` sino que simplemente escribía None, ahora definiendo mi_sonido como abstracto (decorandolo con @abstractmehod) cuando queremos colocar Alien("Predator") en la lista jauria nos va a dar la excepción

`TypeError: Can't instantiate abstract class Alien without an implementation for abstract method 'mi_sonido' ` ya que

_no se puede crear una instancia de una clase marcada como abstracta porque tiene al menos un método abstracto_

```python
from abc import ABC, abstractmethod

class Animal(ABC):
    def __init__(self, nombre: str):
        self.nombre = nombre

    def __str__(self):
        return f"Soy un {type(self).__name__} de nombre {self.nombre} que hace {self.mi_sonido()}"

    @abstractmethod
    def mi_sonido(self) -> str: pass #Animal y los que hereden de Animal deben implementar este método para poder crear instancias


#Al no definir un __init__ se llama implicitamente al de la clase base (super)
class Perro(Animal):
    def mi_sonido(self):
        return "Jau, Jau"

class Gato(Animal):
    def mi_sonido(self):
        return "Miau, Miau"

class Pato(Animal):
    def mi_sonido(self):
        return "Cuac, Cuac"

class Leon(Animal):
    def mi_sonido(self):
        return "Gouarr, Gouarr"

class Alien(Animal):
    pass

pe = Perro("Pluto")
print(pe)
g = Gato("Tom")
print(g)
pa = Pato("Donald")
print(pa)
l = Leon("Simba")
print(l)
jauria = [pe, g, pa, l]
a_sonar(jauria)
jauria.append(Alien("Predator")) #Debe dar error al instanciar Alien
a_sonar(jauria)

```

## 4.2 Ejemplo Jerarquía de personas
El siguiente ejemplo nos muestra una jerarquía de personas. A nivel de persona podemos decir que tiene nombre, fecha de nacimiento y una edad (que implementamos como un método que calcula la edad a partir de la fecha de nacimiento y la fecha actual

```python
from datetime import datetime, date

class Persona:
    def __init__(self, nombre: str, dia: int, mes: int, año: int):
        self.nombre = nombre
        self.fecha_nacimiento = date(año, mes, dia)

    def edad(self) -> int:
        hoy = datetime.now().date()
        edad = hoy.year - self.fecha_nacimiento.year
        if (hoy.month, hoy.day) < (self.fecha_nacimiento.month, self.fecha_nacimiento.day):
            #estamos en una fecha anterior a la de mi cumpleaños
            edad -= 1
        return edad
    def __str__(self):
        return f"Soy {self.nombre} tengo {self.edad()} años"
    

yo = Persona("Miguel", 18, 3, 1949)
print(yo)
rocio = Persona("Rocio", 31, 1, 2009)
print(rocio)
```

Podemos definir ahora tres clases que heredan de `Persona` y que tipifican a personas según sus ingresos: `Estudiante`, `Trabajador`,` Jubilado`. Para crear una instancia de estas nuevas clases tenemos que dar un dato más ingresos además de los datos que se dan para todas persona. El código `__init__` además de inicializar las mismas variables que inicializaba para Persona debe inicializar la nueva variable ingresos que ahora tienen las nuevas clases que heredan de `Persona`.

Para no volver a reescribir eso en cada una de las clases herederas lo que se puede hacer en la clase heredera es _invocar a un método de la clase de la cual hereda._ mediante el recurso `super` Observe en el código de la celda siguiente como se hace

 `super().__init__(nombre, dia, mes, año) ` y luego se inicializan las nuevas variables, en este ejemplo

`self.ingresos = ingresos`

Observe también la refinición del método `__str__`, se llama a `super().__str__()` para que este ponga el nombre y la edad y luego este pone el valor de ingresos.

Este es un patrón típico de la herencia. _Los métodos que se redefinen con la herencia suelen hacer lo mismo que el método correspondiente de la clase base (de ahí la facilidad para llamar a_` super`) _más algo adicional que es propio de la nueva clase_

```python
class Estudiante(Persona):
    def __init__(self, nombre: str, dia: int, mes: int, año: int, ingresos: float = 0.0):
        super().__init__(nombre, dia, mes, año)
        self.ingresos = ingresos
    def __str__(self):
        return f"{super().__str__()}. Soy estudiante mis ingresos son {self.ingresos} CUPs"

class Trabajador(Persona):
    def __init__(self, nombre: str, dia: int, mes: int, año: int, ingresos: float):
        super().__init__(nombre, dia, mes, año)
        self.ingresos = ingresos
    def __str__(self):
        return f"{super().__str__()}. Mi salario es de {self.ingresos} CUPs"

class Jubilado(Persona):
    def __init__(self, nombre: str, dia: int, mes: int, año: int, ingresos: float):
        super().__init__(nombre, dia, mes, año)
        self.ingresos = ingresos
    def __str__(self):
        return f"{super().__str__()}. Mi jubilación es de {self.ingresos} CUPs"

t = Trabajador("Juan", 18, 3, 1979,10000)
e = Estudiante("Rocío", 31, 1, 2009)
j = Jubilado("Luis", 31, 12, 1960, 4000)
print(t)
print(j)
print(e)

```

## 4.3 Jerarquía de Figura

La jerarquía a continuación nos muestra una clase base `Figura`. Esta clase se define como **abstracta** porque no sabemos realmente de que tipo de figuras se está hablando, pero suponemos que sean fuguras de las cuales podamos hablar de su _perimetro_ y su _área_. Se podría también establecer una comparación de orden entre figuras basada en el perímetro o en su área.

Por eso los métodos `perimetro` y `area` en la definición a continuación son abstractos. Sin embargo, basándonos en los objetos que se puedan crear de tipos que hereden de `Figura` tienen que implementar `perimetro` y `area` entonces se puede redefinir la función `__lt__` con la comparación

`self.perimetro()<f.perimetro()`

```python
from typing import List
from abc import ABC, abstractmethod

class Figura(ABC):
    @abstractmethod
    def perimetro(self) -> float:
        pass

    @abstractmethod
    def area(self) -> float:
        pass

    def __lt__(self, f) -> bool:
        if not isinstance(f, Figura):
          raise TypeError(f"No se puede comparar con < una Figura con un {type(f).__name__}")
        return self.perimetro() < f.perimetro()
```

Las clases `Poligono` y `Circulo` (a continuación) heredan de `Figura`. Note que en el `__init__` de ambas clases no tiene que llamar a ningún `super().__init__ `porque no hay un `__init__` común en la clase base `Figura`. Sin embargo, la función `__lt__` implementada en la clase base `Figura` se adaptará a la correspondiente definición de la función `perimetro `en las clases `Circulo` y `Poligono`

```python
import math
class Punto:
    def __init__(self, x: int, y: int) -> None:
        self.x: int = x
        self.y: int = y

    def __str__(self) -> str:
        return f"({self.x}, {self.y})"

    def distancia(self, otro_punto: 'Punto') -> float:
        dx: float = otro_punto.x - self.x
        dy: float = otro_punto.y - self.y
        return math.sqrt(dx**2 + dy**2)

class Poligono(Figura):
    def __init__(self, vertices: List[Punto]):
        self.vertices:List[Punto] = vertices
        if len(self.vertices) < 3:
            raise ValueError(
              f"Un polígono necesita al menos 3 puntos, recibido: {len(self.vertices)}"
            )
    def __str__(self) -> str:
        return "Poligono de vértices [" + ",".join(str(v) for v in self.vertices) + "]"

    def perimetro(self) -> float:
        total: float = 0.0
        n: int = len(self.vertices)

        for i in range(n):
            p1: Punto = self.vertices[i]
            p2: Punto = self.vertices[(i + 1) % n]
            total += p1.distancia(p2)
        return total

    def area(self) -> float:
        raise NotImplemented("area por implementar")

class Circulo(Figura):
    def __init__(self, centro:Punto, radio:float) -> float:
        self.centro = centro
        if radio < 0:
            raise ValueError("Radio debe ser > 0")
        self.radio = radio

    def __str__(self) -> str:
        return f"Circulo centro {self.centro} radio {self.radio}"

    def perimetro(self) -> float:
        return 2*(math.pi)*self.radio

    def area(self) -> float:
        return math.pi*(self.radio**2)

circ1 = Circulo(Punto(100,100), 2)

p1: Punto = Punto(0, 0)
p2: Punto = Punto(3, 0)
p3: Punto = Punto(3, 4)
p4: Punto = Punto(0, 4)
pol1 = Poligono = Poligono([p1, p2, p3, p4])

print(circ1)
print(pol1)

if circ1 < pol1:
    print(f"Circulo de perimetro {circ1.perimetro()} es menor que Poligono de perimetro {pol1.perimetro()}")
else:
    print(f"Poligono de perimetro {pol1.perimetro()} es menor que Circulo de perimetro {circ1.perimetro()}")
```
